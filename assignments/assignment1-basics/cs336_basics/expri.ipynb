```python
import torch
from einops import rearrange, einsum

images = torch.randn(64, 128, 128, 3)  # Example input tensor with shape (batch_size, channels, height, width)
dim_by = torch.linspace(start=0.0, end=1.0, steps=10)

dim_value = rearrange(dim_by, "dim_value -> 1 dim_value 1 1 1")
images_rearr = rearrange(images, "b h w c -> b 1 h w c")

dimmed_image = images_rearr * dim_value

dimmed_image_2 = einsum(images, dim_by, "b h w c, dim_value -> b dim_value h w c")

assert torch.allclose(dimmed_image, dimmed_image_2), "两种方法结果不一致！"
print("成功！两种方法结果一致。")
```
